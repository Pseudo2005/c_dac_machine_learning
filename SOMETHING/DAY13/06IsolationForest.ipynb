```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import seaborn as sns
```

```python
df = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\milk.csv", index_col = 0)
clf = IsolationForest( contamination= 0.05, random_state = 26)
clf.fit(df)
predictions = clf.predict(df)
print('%age of outliers = ' + str((predictions<0).mean()*100)+ '%')
abd_ind = np.where(predictions < 0)
print("Outliers")
print(df.index[abd_ind])      
```
