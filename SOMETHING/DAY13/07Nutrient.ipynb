```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
```

```python
nutrient = pd.read_csv(r"F:\MACHINE LEARNING\SOMETHING\Datasets\nutrient.csv", index_col = 0)
clf = IsolationForest(contamination = 0.05, random_state = 26)
clf.fit(nutrient)
prediction = clf.predict(nutrient)
print('%age of outliers = ' + str((prediction<0).mean()*100)+ '%')
abd_ind = np.where(prediction < 0)
print("Outliers")
print(nutrient.index[abd_ind])      
```
