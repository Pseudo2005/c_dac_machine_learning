```python
import surprise
import numpy as np
import pandas as pd
from surprise.model_selection import train_test_split
from surprise.model_selection import GridSearchCV
from surprise.model_selection.split import KFold
```

```python
df = pd.read_csv("F:\MACHINE LEARNING\Cases_rec_sys\Amazon Movie Ratings\Amazon.csv")
ratings = pd.melt(df, id_vars='user_id', var_name="item_id",value_name="rating")
ratings = ratings[ratings['rating'].notna()]
```

```python
ratings.head(20)
```

```python
lowest_rating = ratings['rating'].min()
highest_rating=ratings['rating'].max()
lowest_rating,highest_rating
```

```python
reader = surprise.Reader(rating_scale = (lowest_rating, highest_rating))
data = surprise.Dataset.load_from_df(ratings, reader)
type(data)
```

```python
similarity_options = {'name': 'cosine', 'user_based': True}
algo = surprise.KNNBasic(sim_options = similarity_options) # default k = 40
output = algo.fit(data.build_full_trainset())
# calculates the expected ratings for all the users
```

```python
param_grid = {'n_epochs': [50], 'n_factors':[100,50],
              'lr_all': np.linspace(0.001,1,5),
              'reg_all': np.linspace(0.01,0.8,5)
             }
kfold = KFold(n_splits = 5, random_state = 26, shuffle = True)
gs = GridSearchCV(surprise.SVD, param_grid , measures = ['rmse','mae'], cv = kfold)
gs.fit(data)
```

```python
print(gs.best_score['rmse'])
print(gs.best_params['rmse'])
```

```python
param_grid = {'k': [25,40,50,65,75,85], 'user_bases': [False, True]}
kfold = KFold(n_splits = 5, random_state = 26, shuffle = True)
gs = GridSearchCV(surprise.KNNWithZScore, param_grid , measures = ['rmse','mae'], cv = kfold)
gs.fit(data)
```

```python
print(gs.best_score['rmse'])
print(gs.best_params['rmse'])
```

```python
param_grid = {'k': [25,40,50,65,75,85], 'user_bases': [False, True]}
kfold = KFold(n_splits = 5, random_state = 26, shuffle = True)
gs = GridSearchCV(surprise.KNNBasic, param_grid , measures = ['rmse','mae'], cv = kfold)
gs.fit(data)
```
