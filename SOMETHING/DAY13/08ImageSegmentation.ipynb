```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import StratifiedKFold,GridSearchCV
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings('ignore')
```

```python
img = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Cases\Image_Segmentation\Image_Segmentation.csv")
X,y = img.drop('Class', axis = 1), img.Class
# clf = IsolationForest( contamination= 0.05, random_state = 26)
# clf.fit(df)
# predictions = clf.predict(df)
# print('%age of outliers = ' + str((predictions<0).mean()*100)+ '%')
# abd_ind = np.where(predictions < 0)
# print("Outliers")
# print(df.index[abd_ind])      
```

```python
kfold = StratifiedKFold(n_splits=5,  shuffle=True, random_state=26)
lr = LogisticRegression()
params = {'C': np.linspace(0.01,5,10),
          'solver': ['lbfgs','newton-cg', 'newton-cholesky', 'sag', 'saga']}
gcv = GridSearchCV(lr, param_grid= params, cv = kfold, scoring = 'neg_log_loss')
gcv.fit(X,y)
print(gcv.best_params_, gcv.best_score_)
```

```python
clf = IsolationForest( contamination= 0.05, random_state = 26)
clf.fit(X)
predictions = clf.predict(X)
print('%age of outliers = ' + str((predictions<0).mean()*100)+ '%')
abd_ind = np.where(predictions < 0)[0]
print("Outliers")
print(X.index[abd_ind]) 
```

```python
X.drop(abd_ind,inplace=True)
y.drop(abd_ind,inplace=True)
```

```python
kfold = StratifiedKFold(n_splits=5,  shuffle=True, random_state=26)
lr = LogisticRegression()
params = {'C': np.linspace(0.01,5,10),
          'solver': ['lbfgs','newton-cg', 'newton-cholesky', 'sag', 'saga']}
gcv = GridSearchCV(lr, param_grid= params, cv = kfold, scoring = 'neg_log_loss')
gcv.fit(X,y)
print(gcv.best_params_, gcv.best_score_)
```
