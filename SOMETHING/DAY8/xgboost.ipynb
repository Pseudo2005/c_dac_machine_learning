```python
from sklearnex import patch_sklearn
patch_sklearn()
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import accuracy_score, r2_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier, GradientBoostingRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.pipeline import Pipeline
from tqdm import tqdm
```

```python
hr = pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\human-resources-analytics\HR_comma_sep.csv")
X,y=hr.drop('left',axis=1),hr['left']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=.3,stratify=y)
hr.head()
```

```python
ohe=OneHotEncoder(sparse_output=False,drop='first').set_output(transform='pandas')
ct=make_column_transformer((ohe,make_column_selector(dtype_include=object)),
                           ('passthrough',make_column_selector(dtype_exclude=object)), verbose_feature_names_out=False ).set_output(transform='pandas')
ct=ct.set_output(transform='pandas')
ct_ohe_tr=ct.fit_transform(X_train)
ct_ohe_tst=ct.transform(X_test)
```

```python

learn_r = np.linspace(0.01, 0.9, 10)
max_depth = [None, 2,3,4]
n_est = [25,50,100]
scores = []
for r in tqdm(learn_r):
    for m in max_depth:
        for n in n_est:
            gbm = GradientBoostingClassifier(random_state=26, learning_rate=r, max_depth= m, n_estimators=n)
            gbm.fit(ct_ohe_tr, y_train)
            pred = gbm.predict(ct_ohe_tst)
            scores.append([r,m,n, accuracy_score(y_test, pred)])
score_df=pd.DataFrame(scores,columns=["learning_rate",'max_depth','n_estimators','accuracy_score(' ])
score_df.sort_values(by="accuracy_score(",ascending=False,ignore_index=True)[:20]
```
