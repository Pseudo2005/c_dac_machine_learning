```python
from sklearnex import patch_sklearn
patch_sklearn()
import pandas as pd

from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import accuracy_score, r2_score

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.pipeline import Pipeline
from tqdm import tqdm
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\Vehicle Silhouettes\Vehicle.csv")
df.head()
```

```python
X,y=df.drop("Class",axis=1),df.Class
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26,stratify=y)
```

```python
sc=StandardScaler().set_output(transform='pandas')
learn_r = np.linspace(0.01, 0.9, 10)
max_depth = [None, 2,3,4]
n_est = [25,50,100]
scores = []
for r in tqdm(learn_r):
    for m in max_depth:
        for n in n_est:
            gbm = GradientBoostingClassifier(random_state=26, learning_rate=r, max_depth= m, n_estimators=n)
            pipe=Pipeline([('sc',sc),('gbm',gbm)])
            pipe.fit(X_train, y_train)
            pred = pipe.predict(X_test)
            scores.append([r,m,n, accuracy_score(y_test, pred)])
score_df=pd.DataFrame(scores,columns=["learning_rate",'max_depth','n_estimators','accuracy_score(' ])
score_df.sort_values(by="accuracy_score(",ascending=False,ignore_index=True)[:20]


```

```python
from catboost  import CatBoostClassifier
learn_r = np.linspace(0.01, 0.9, 10)
max_depth = [None, 2,3,4]
n_est = [25,50,100]
scores = []
for r in tqdm(learn_r):
    for m in max_depth:
        for n in n_est:
            gbm = CatBoostClassifier(random_state=26, learning_rate=r, max_depth= m, n_estimators=n, verbose = False)
            gbm.fit(X_train, y_train)
            pred = gbm.predict(X_test)
            scores.append([r,m,n, accuracy_score(y_test, pred)])
score_df=pd.DataFrame(scores,columns=["learning_rate",'max_depth','n_estimators','accuracy_score(' ])
score_df.sort_values(by="accuracy_score(",ascending=False,ignore_index=True)[:20]
```
