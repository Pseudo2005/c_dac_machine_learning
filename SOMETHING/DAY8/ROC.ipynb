```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve,roc_auc_score,RocCurveDisplay
```

```python
df = pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Datasets\Datasets\comp_prob.csv")
y_test, yprob_1, yprob_2 = df['y_test'], df['yprob_1'], df['yprob_2']
df.head()
```

```python
fpr , tpr, thres = roc_curve(y_test, yprob_1)
plt.scatter(fpr, tpr, c = 'red')
plt.plot(fpr, tpr)
plt.xlabel('1-Specifity')
plt.ylabel('Sensitivity')
plt.show()
```

```python
fpr , tpr, thres = roc_curve(y_test, yprob_1)
auc_score=roc_auc_score(y_test,yprob_1)
dis=RocCurveDisplay(fpr=fpr,tpr=tpr,roc_auc=auc_score)
dis.plot()

plt.show()
```
