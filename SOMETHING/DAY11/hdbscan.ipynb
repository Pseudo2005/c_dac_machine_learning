```python
import numpy as np
import pandas as pd
from sklearn.cluster import HDBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
milk = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\milk.csv", index_col = 0)
milk.head()
```

```python
scaler=StandardScaler().set_output(transform='pandas')
scl_milk=scaler.fit_transform(milk)
```

```python
hdbs = HDBSCAN(min_cluster_size=3)
hdbs.fit(scl_milk)
inliers_obs=scl_milk[hdbs.labels_ != -1]
inliers_lbls=hdbs.labels_[hdbs.labels_ != -1]
silhouette_score(inliers_obs,inliers_lbls)
```
