```python
import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
milk = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\milk.csv", index_col = 0)
milk.head()
```

```python

scaler=StandardScaler().set_output(transform='pandas')
scl_milk=scaler.fit_transform(milk)
dbs=DBSCAN(eps=1.5,min_samples=2)
dbs.fit(scl_milk)
inliers_obs=scl_milk[dbs.labels_ != -1]
inliers_lbls=dbs.labels_[dbs.labels_ != -1]
silhouette_score(inliers_obs,inliers_lbls)
```

```python
eps=np.linspace(.1,3,12)
minp=[2,3,4,5]
score=[]
for e in eps:
    for m in minp:
        dbs=DBSCAN(eps=e,min_samples=m)
        dbs.fit(scl_milk)
        inliers_obs=scl_milk[dbs.labels_ != -1]
        inliers_lbls=dbs.labels_[dbs.labels_ != -1]

        score.append([e,m,silhouette_score(inliers_obs,inliers_lbls)])
db_score=pd.DataFrame(score,columns=['eps','min_sample','score']).sort_values(by='score',ascending=False)
```
