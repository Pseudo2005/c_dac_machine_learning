```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
from sklearn.metrics import accuracy_score
from sklearn.ensemble import VotingClassifier
```

```python
heart = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\heart.csv")
X,y=heart.drop('target',axis=1),heart.target
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26,stratify=y)
```

```python
dtc=DecisionTreeClassifier(random_state=26)
lr=LogisticRegression()
knn=KNeighborsClassifier()
voting=VotingClassifier([('TREE',dtc),('LR',lr),('KNN',knn)])
voting.fit(X_train,y_train)
y_pred=voting.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\Wisconsin\BreastCancer.csv",index_col=0)
X,y=df.drop('Class',axis=1),df['Class']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=.3,stratify=y)
df.head()
```

```python
dtc=DecisionTreeClassifier(random_state=26)
lr=LogisticRegression()
knn=KNeighborsClassifier()
voting=VotingClassifier([('TREE',dtc),('LR',lr),('KNN',knn)])
voting.fit(X_train,y_train)
y_pred=voting.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
depths=[None,2,3,4,5]
solvers=['lbfgs','saga','sag','liblinear']
Ks=[2,5,10]
scores=[]
for d in tqdm(depths):
    for s in solvers:
        for k in Ks:
            dtc=DecisionTreeClassifier(random_state=26,max_depth=d)
            lr=LogisticRegression(solver=s)
            knn=KNeighborsClassifier(n_neighbors=k)
            voting=VotingClassifier([('TREE',dtc),('LR',lr),('KNN',knn)],voting='hard')
            voting.fit(X_train,y_train)
            y_pred=voting.predict(X_test)
            scores.append([d,s,k,accuracy_score(y_test,y_pred)])
pd.DataFrame(scores,columns=['DEPTH','SOLVER','N_NEIGHBORS','ACCURACY']).sort_values(by='ACCURACY',ascending=False)
            
```

```python
depths=[None,2,3,4,5]
solvers=['lbfgs','saga','sag','liblinear']
Ks=[2,5,10]
scores=[]
for d in tqdm(depths):
    for s in solvers:
        for k in Ks:
            dtc=DecisionTreeClassifier(random_state=26,max_depth=d)
            lr=LogisticRegression(solver=s)
            knn=KNeighborsClassifier(n_neighbors=k)
            voting=VotingClassifier([('TREE',dtc),('LR',lr),('KNN',knn)],voting='soft')
            voting.fit(X_train,y_train)
            y_pred=voting.predict(X_test)
            scores.append([d,s,k,accuracy_score(y_test,y_pred)])
pd.DataFrame(scores,columns=['DEPTH','SOLVER','N_NEIGHBORS','ACCURACY']).sort_values(by='ACCURACY',ascending=False)
            
```
