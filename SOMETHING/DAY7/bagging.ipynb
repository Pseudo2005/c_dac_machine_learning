```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
from sklearn.metrics import accuracy_score,r2_score
from sklearn.ensemble import BaggingClassifier,BaggingRegressor
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\Wisconsin\BreastCancer.csv",index_col=0)
X,y=df.drop('Class',axis=1),df['Class']
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=26,test_size=.3,stratify=y)
df.head()
```

```python
lr=LogisticRegression()
knn=KNeighborsClassifier()
dtc=DecisionTreeClassifier(random_state=26)
bag=BaggingClassifier(estimator=lr,n_estimators=15,random_state=26)
bag.fit(X_train,y_train)
pred=bag.predict(X_test)
accuracy_score(y_test,pred)
```

```python
est = [lr,knn, dtc]
n_samples = [10,25,50]
scores = []
for e in est:
    for n in n_samples:
        bag = BaggingClassifier(estimator = e, n_estimators = n, random_state=26)
        bag.fit(X_train, y_train)
        pred = bag.predict(X_test)
        scores.append([e, n, accuracy_score(y_test, pred)])
df_scores = pd.DataFrame(scores, columns=['Estimator','n_estimator','Score'])
df_scores.sort_values('Score', ascending = False)
```

```python
from sklearn.linear_model import Ridge,Lasso,LinearRegression
from sklearn.tree import DecisionTreeRegressor
concrete = pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\Concrete_Strength\Concrete_Data.csv")
X,y=concrete.drop('Strength',axis=1),concrete.Strength
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26)
```

```python
ridge=Ridge()
lasso=Lasso()
lr=LinearRegression()
dtr=DecisionTreeRegressor(random_state=26)
est=[ridge,lasso,lr,dtr]
n_sample=[10,15,20,25,40,50]
scores=[]
for e in est:
    for n in n_sample:
        bag=BaggingRegressor(estimator=e,n_estimators=n,random_state=26)
        bag.fit(X_train,y_train)
        pred=bag.predict(X_test)
        scores.append([e,n,r2_score(y_test,pred)])
pd.DataFrame(scores,columns=["Estimator","n_estimator","score"]).sort_values("score",ascending=False,ignore_index=True)
```
