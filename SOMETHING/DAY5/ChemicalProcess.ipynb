```python
import pandas as pd
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.metrics import r2_score
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\ChemicalProcess.csv")
df.head()
```

```python
df.info()
```

```python
df=df.fillna(value=df.mean())
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
X=df.drop("Yield",axis=1)
y=df.Yield
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=26)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
pred=lr.predict(X_test)
r2_score(y_test,pred)
```

```python
ridge=Ridge()
ridge.fit(X_train,y_train)
pred=ridge.predict(X_test)
r2_score(y_test,pred)
```

```python
lsso=Lasso()
lsso.fit(X_train,y_train)
pred=lsso.predict(X_test)
r2_score(y_test,pred)
```

```python
els=ElasticNet()
els.fit(X_train,y_train)
pred=els.predict(X_test)
r2_score(y_test,pred)
```
