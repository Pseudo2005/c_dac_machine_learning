```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.model_selection import train_test_split,KFold,GridSearchCV
from sklearn.metrics import r2_score
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Cases\Chemical Process Data\ChemicalProcess.csv")
X,y=df.drop("Yield",axis=1),df.Yield

df.head()
```

```python
imp=SimpleImputer()
mm=MinMaxScaler()
ss=StandardScaler()
knn=KNeighborsRegressor()
kf=KFold(n_splits=5,shuffle=True, random_state=26)
pipe=Pipeline([('IMP',imp),('SCL',ss),('KNN',knn)])
```

```python
pipe.get_params()
```

```python
params={'IMP__strategy': ['mean','median'],'SCL':[ss,mm,None],'KNN__n_neighbors':np.arange(1,11)}
gcv=GridSearchCV(pipe,param_grid=params,cv=kf,scoring='r2')
gcv.fit(X,y)
print(gcv.best_params_)
print(gcv.best_score_)
```
