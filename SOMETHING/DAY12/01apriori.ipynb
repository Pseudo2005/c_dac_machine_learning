```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
import pandas as pd
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\Faceplate.csv")
df = df.astype(bool)
ap = apriori(df, min_support= 0.2,  use_colnames=True)
ap
```

```python
rules = association_rules(ap, metric = 'confidence', min_threshold=0.5)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by = ['confidence','lift','support'], ascending = False)
```

```python
relv_rule = rules[rules['lift']>1]
relv_rule.sort_values(by =['confidence','lift','support'], ascending = False)
```

```python
cosmetic = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\Cosmetics.csv", index_col=0)
cosmetic.head()
```

```python
cos = cosmetic.astype(bool)
ap_cos = apriori(cos, min_support= 0.2,  use_colnames=True)
ap_cos
```

```python
rules = association_rules(ap_cos, metric = 'confidence', min_threshold=0.5)
rules = rules[['antecedents','consequents','support','confidence','lift']]
rules.sort_values(by = ['confidence','lift','support'], ascending = False)
```

```python
from ipywidgets import interact
import ipywidgets as widgets
```

```python
def gen_rules(min_support, min_confidence):
    itemsets = apriori(cos, min_support= min_support, use_colnames = True)
    rules = association_rules(itemsets, metric = 'confidence', min_threshold= min_confidence)
    rules = rules[['antecedents','consequents','support','confidence','lift']]
    relv_rules = rules[rules['lift'] > 1]
    return relv_rules.sort_values(by = ['confidence', 'lift', 'support'], ascending = False)
```

```python
interact(gen_rules,
         min_support = widgets.FloatSlider( value = 0.01, min = 0, max = 1, step = 0.01, description = 'Min Support: '),
         min_confidence = widgets.FloatSlider( value = 0.01, min =0, max = 1, step = 0.01, description = 'Min confidence'))
```
