```python
import pandas as pd 
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.metrics import accuracy_score
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.pipeline import Pipeline
import numpy as np
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\student_performance\student_dataset_10000_rows.csv")
X,y=df.drop('placement_status',axis=1),df.placement_status
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26,stratify=y)
```

```python
svm = SVC(kernel='linear')
svm.fit(X_train, y_train)
pred = svm.predict(X_test)
accuracy_score(y_test, pred)
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\human-resources-analytics\HR_comma_sep.csv")
X,y=df.drop('left',axis=1),df.left
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26)
```

```python
ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
sc=StandardScaler().set_output(transform='pandas')
ct=make_column_transformer((ohe,make_column_selector(dtype_include=object)),('passthrough',make_column_selector(dtype_exclude=object)),verbose_feature_names_out=False)
svm = SVC(kernel='linear')
pipe=Pipeline([('ct',ct),('sc',sc),('svm',svm)])
pipe.fit(X_train, y_train)
pred = pipe.predict(X_test)
accuracy_score(y_test, pred)
```

```python
ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
sc=StandardScaler().set_output(transform='pandas')
ct=make_column_transformer((ohe,make_column_selector(dtype_include=object)),('passthrough',make_column_selector(dtype_exclude=object)),verbose_feature_names_out=False)
cs=np.linspace(.001,10,20)
scores=[]
for c in tqdm(cs):
    svm = SVC(kernel='linear',C=c)
    pipe=Pipeline([('ct',ct),('sc',sc),('svm',svm)])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_test)
    scores.append([c,accuracy_score(y_test, pred)])

df_score=pd.DataFrame(scores,columns=['C','scores'])
df_score.sort_values('scores',ascending=False,ignore_index=True)[:20]    
    
```

```python
ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
sc=StandardScaler().set_output(transform='pandas')
ct=make_column_transformer((ohe,make_column_selector(dtype_include=object)),('passthrough',make_column_selector(dtype_exclude=object)),verbose_feature_names_out=False)
cs=np.linspace(.001,10,20)
gs = np.linspace(.01,10,20)
scores=[]
for c in tqdm(cs):
    for g in gs:
        svm = SVC(kernel='rbf',C=c,gamma=g)
        pipe=Pipeline([('ct',ct),('sc',sc),('svm',svm)])
        pipe.fit(X_train, y_train)
        pred = pipe.predict(X_test)
        scores.append([c,g,accuracy_score(y_test, pred)])
df_score=pd.DataFrame(scores,columns=['C','gamma','scores'])
df_score.sort_values('scores',ascending=False,ignore_index=True)[:20]  
```
