```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.compose import make_column_selector,make_column_transformer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\human-resources-analytics\HR_comma_sep.csv")
df.head()
```

```python
X,y=df.drop('left',axis=1),df.left
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26)
```

```python
ohe = OneHotEncoder(sparse_output= False).set_output(transform = 'pandas')
ct = make_column_transformer((ohe, make_column_selector(dtype_include=object)),
                            ('passthrough',make_column_selector(dtype_include = ['int64','float64'])),
                            verbose_feature_names_out=False)
ct=ct.set_output(transform='pandas')
scaler=StandardScaler()
```

```python
solvers=['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']
scores=[]
for s in solvers:
    lr = LogisticRegression(solver=s)
    pipe=Pipeline([("ct",ct),('Scaler',scaler),("LR",lr)])
    pipe.fit(X_train,y_train)
    pred=pipe.predict(X_test)
    scores.append([s,accuracy_score(y_test, pred)])
df_score=pd.DataFrame(scores,columns=['solver','scores'])
df_score.sort_values('scores',ascending=False,ignore_index=True)    
    
```
