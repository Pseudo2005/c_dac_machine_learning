```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("F:\MACHINE LEARNING\SessionsMaterial\Cases\student_performance\student_dataset_10000_rows.csv")
X,y = df.drop('placement_status', axis = 1), df.placement_status
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.3,random_state=26,stratify=y)
```

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)
pred = lr.predict(X_test)
accuracy_score(y_test, pred)
```
