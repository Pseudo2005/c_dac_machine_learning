```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\monthly-milk-production-pounds-p.csv",index_col=0)
df.index=pd.to_datetime(df.index).to_period('M')
df.index
```

```python
df.plot()
plt.title("Monthly milk production")
plt.xlabel('Month')
plt.show()
```

```python
downsample = df.resample('Q').sum()
downsample.index.rename('Quarter', inplace = True)
downsample.plot()
plt.title("Quarterly milk production")
plt.xlabel('Quarter')
plt.show()
```
