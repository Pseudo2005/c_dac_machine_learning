```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
from pca import pca
```

```python
nutrient=pd.read_csv(r"F:\MACHINE LEARNING\SOMETHING\Datasets\nutrient.csv",index_col=0)
nutrient.head()
```

```python
model=pca(normalize=True)
res=model.fit_transform(nutrient,row_labels=nutrient.index)
```

```python
sc=StandardScaler().set_output(transform='pandas')
s_nut=sc.fit_transform(nutrient)
pcaa=PCA().set_output(transform='pandas')
p_nut=pcaa.fit_transform(s_nut)
```

```python
model.biplot()
for n in range(len(nutrient)):
    plt.text(p_nut.pca0.iloc[n],p_nut.pca1.iloc[n],p_nut.index[n])

    
```
