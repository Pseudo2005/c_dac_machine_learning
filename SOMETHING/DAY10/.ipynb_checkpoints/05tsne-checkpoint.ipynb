```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE
```

```python
wisc = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Cases\Wisconsin\BreastCancer.csv", index_col = 0)
X, y = wisc.drop('Class', axis = 1), wisc.Class
tsne = TSNE(random_state=26).set_output(transform = 'pandas')
t_comps = tsne.fit_transform(X)
```

```python
t_comps['Class'] = y
sns.scatterplot(x = 'tsne0', y='tsne1', hue = 'Class', data= t_comps)
plt.show()
```
