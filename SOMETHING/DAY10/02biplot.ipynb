```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
from pca import pca
import warnings
warnings.filterwarnings('ignore')
```

```python
milk = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\milk.csv", index_col = 0)
model = pca(normalize = True)
results = model.fit_transform(milk, row_labels = milk.index)
```

```python
scaler = StandardScaler().set_output(transform = 'pandas')
scl_milk = scaler.fit_transform(milk)

prcomp = PCA().set_output(transform = 'pandas')
pca_data = prcomp.fit_transform(scl_milk)
```

```python
model.biplot()
for n in range(len(milk)):
    plt.text(pca_data['pca0'].iloc[n],pca_data.pca1.iloc[n],milk.index[n])
    
plt.show()
```
