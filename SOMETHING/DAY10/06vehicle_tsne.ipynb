```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.manifold import TSNE
```

```python
vehicle = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Cases\Vehicle Silhouettes\Vehicle.csv")
X,y = vehicle.drop('Class', axis = 1), vehicle.Class

vehicle.head()
```

```python
tsn=TSNE(random_state=26,perplexity=20).set_output(transform='pandas')
scale = StandardScaler()
scl_x = scale.fit_transform(X)
ts_ve=tsn.fit_transform(scl_x)
ts_ve['Class']=y
sns.scatterplot(data=ts_ve,x='tsne0',y='tsne1',hue='Class')
```
