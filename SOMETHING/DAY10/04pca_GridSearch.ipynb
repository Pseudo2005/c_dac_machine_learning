```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
from pca import pca
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
import warnings
warnings.filterwarnings('ignore')
```

```python
vehicle = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Cases\Vehicle Silhouettes\Vehicle.csv")
X,y = vehicle.drop('Class', axis = 1), vehicle.Class
# X_train, X_test, y-train, y_test = train_test_split(X,y, test_size=0.3, 
vehicle.head()
```

```python
sc = StandardScaler().set_output(transform='pandas')
lr = LogisticRegression(random_state=26)
pca=PCA().set_output(transform='pandas')
kfold=StratifiedKFold(shuffle=True,random_state=26)
pipe=Pipeline([("sc",sc),('pca',pca),("lr",lr)])
```

```python
pipe.get_params()
```

```python
params={
    'lr__C':np.linspace(.001,5,10),
    'pca__n_components':np.arange(2,18)
}
grid=GridSearchCV(pipe,cv=kfold,param_grid=params,scoring='accuracy',n_jobs=-1)
grid.fit(X,y)
print(grid.best_params_)
print(grid.best_score_)
```

```python
svc=SVC( kernel='linear',probability=True, random_state=26)
pipe1=Pipeline([("sc",sc),('pca',pca),("svc",svc)])
pipe1.get_params()
```

```python
params={
    'svc__C':np.linspace(.001,5,10),
    'pca__n_components':np.arange(2,18)
    
}
grid=GridSearchCV(pipe1,cv=kfold,param_grid=params,scoring='neg_log_loss',n_jobs=-1)
grid.fit(X,y)
print(grid.best_params_)
print(grid.best_score_)
```
