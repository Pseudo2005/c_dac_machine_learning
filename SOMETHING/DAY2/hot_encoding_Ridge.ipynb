```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error
from sklearn.preprocessing import PolynomialFeatures, OneHotEncoder
from sklearn.linear_model import LinearRegression,Ridge
```

```python
df = pd.read_csv("F:\MACHINE LEARNING\SOMETHING\Datasets\Housing.csv")
df.isnull().sum()
```

```python
X,y = df.drop('price', axis = 1), df['price']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=26)
```

```python
ohe=OneHotEncoder(sparse_output=False).set_output(transform='pandas')
col_ohe_train=X_train.select_dtypes(include='str')
OHE_train=ohe.fit_transform(col_ohe_train)
col_ohe_test=X_test.select_dtypes(include='str')
OHE_test=ohe.transform(col_ohe_test)
```

```python
X_train=pd.concat([OHE_train,X_train],axis=1)
X_test=pd.concat([OHE_test,X_test],axis=1)
X_train
```

```python
X_train.drop(columns=X_train.select_dtypes(include="str").columns,inplace=True)
X_test.drop(columns=X_test.select_dtypes(include="str").columns,inplace=True)
```

```python
lr=LinearRegression()
lr.fit(X_train,y_train)
pred=lr.predict(X_test)
```

```python
root_mean_squared_error(pred,y_test)
```

```python
poly=PolynomialFeatures(degree=2,include_bias=False).set_output(transform='pandas')
X_train1=poly.fit_transform(X_train)
X_test1=poly.transform(X_test)
lr=LinearRegression()
lr.fit(X_train1,y_train)
pred=lr.predict(X_test1)
root_mean_squared_error(pred,y_test)
```

```python
ridge=Ridge(alpha=.9)
ridge.fit(X_train,y_train)
pred=ridge.predict(X_test)
root_mean_squared_error(pred,y_test)
```

```python
alphas=[.1,.25,.7,1,1.5,2,5,10,50]
def rim(x):
    
    ridge=Ridge(alpha=x)
    ridge.fit(X_train,y_train)
    pred=ridge.predict(X_test)
    print(f'at alpha :{x} root mean value: {root_mean_squared_error(pred,y_test)}')
    return root_mean_squared_error(pred,y_test)
root_list=[]
for x in alphas:
    
    root_list.append(rim(x))

root_list    
```
