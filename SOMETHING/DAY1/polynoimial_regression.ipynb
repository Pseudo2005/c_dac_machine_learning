```python
import pandas as pd
import numpy as np 
from sklearn.preprocessing import PolynomialFeatures
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error
import os
os.chdir("F:\MACHINE LEARNING\SOMETHING\Datasets")
```

```python
pizza=pd.read_csv("pizza.csv")
x = pizza[['Promote']]
y = pizza['Sales']
```

```python
poly=PolynomialFeatures(degree=2).set_output(transform='pandas')
X_poly=poly.fit_transform(x)
X_poly
```

```python
insure=pd.read_csv("insure_auto.csv")
X,y=insure.drop(columns=['Operating_Cost','Branch']),insure["Operating_Cost"]
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=34,test_size=0.3)
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```

```python
insure=pd.read_csv("insure_auto.csv")
X,y=insure.drop(columns=['Operating_Cost','Branch']),insure["Operating_Cost"]
poly=PolynomialFeatures(degree=2,include_bias=False).set_output(transform='pandas')
poly_X=poly.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(poly_X,y, random_state=34,test_size=0.3)
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
root_mean_squared_error(y_test,y_pred)
```
